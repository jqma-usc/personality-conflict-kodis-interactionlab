```python
df = pd.read_csv("LIWC_22_Temporal_K-3.csv")
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming the original DataFrame 'df' is already loaded

# Filtering the DataFrame for a specific 'kodis-id'
specific_kodis_id = 2523  # Replace with the desired 'kodis-id'
filtered_df = df[df['kodis-id'] == specific_kodis_id]

# Further filtering for buyers and sellers
buyer_df = filtered_df[filtered_df['buyer or seller'] == 'buyer']
# seller_df = filtered_df[filtered_df['buyer or seller'] == 'seller']

# Define the columns for the plot
x_col = ''
y_col = 'Clout'

# Create the scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(buyer_df[x_col], buyer_df[y_col], label='Buyer', color='blue', alpha=0.7)
# plt.scatter(seller_df[x_col], seller_df[y_col], label='Seller', color='red', alpha=0.7)

plt.xlabel(x_col)
plt.ylabel(y_col)
plt.title(f'Scatter Plot for kodis-id {specific_kodis_id} vs {y_col}')
plt.legend()

# Adjust the y-axis limits according to the data range with a margin
if not filtered_df.empty:
    y_min = buyer_df[y_col].min()
    y_max = buyer_df[y_col].max()
    print(y_max)
    margin = (y_max - y_min) * 0.1  # 10% margin on top and bottom
    plt.ylim(y_min - margin, y_max + margin)

plt.show()
```

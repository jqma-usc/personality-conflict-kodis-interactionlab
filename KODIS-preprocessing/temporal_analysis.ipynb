```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("LIWC_22_Temporal_K-3.csv")

kodis_id = 2867 # TODO turn this into a random number generator
liwc_variable = "polite"

df_plot = df[df["kodis-id"] == kodis_id]
buyer_data = df_plot[df_plot["buyer or seller"] == "buyer"]
seller_data = df_plot[df_plot["buyer or seller"] == "seller"]


assert len(buyer_data) == len(seller_data)
```

```python
plt.plot(np.arange(len(buyer_data)), list(buyer_data[liwc_variable]), color="red")
plt.xticks(np.arange(len(buyer_data)))
plt.plot(np.arange(len(seller_data)), list(seller_data[liwc_variable]), color="blue")
plt.legend(["buyer", "seller"], loc="lower right")
plt.xlabel("Sliding window index")
plt.ylabel(liwc_variable)
# plt.title(f"{liwc_variable} over time for KODIS conversation #{kodis_id}")
plt.title(f"Politeness over time for KODIS conversation #{kodis_id}")
plt.show()
```

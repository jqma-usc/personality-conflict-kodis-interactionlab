# Testing helper functions

```python
kodis_df = pd.read_csv("DO-NOT-DISTRIBUTE-KODIS-human-human-subset.csv")
kodis_df.columns
```

The column "Unnamed: 0" is the kodis-id

```python
conv_num = 23
conv = kodis_df['s_fullChat'][conv_num].replace("<b>","").replace("</b>","").replace("<br>","\n").replace("[Other]","BUYER").replace("[You]","SELLER")
print(conv)
```

Remove the following utterances:
- "I Walk Away"
- "Submitted agreement: ..."
- "Accept Deal"
- "Reject Deal"

These are preset values that the participants can click, and not utterances they type in, so we want to remove them from the LIWC analysis.

```python
"kaleen".startswith("kal")
```

```python
a = "I Walk Away"
b = "Submitted agreement:"
c = "Accept Deal"
d = "Reject Deal"
buyer = "BUYER"
seller = "SELLER"

buyer_turns = []
seller_turns = []

for turn in conv.split("\n")[:-1]:
    if turn[8:].startswith(a) or turn[8:].startswith(b) or turn[8:].startswith(c) or turn[8:].startswith(d):
        continue
    if turn[7:].startswith(a) or turn[7:].startswith(b) or turn[7:].startswith(c) or turn[7:].startswith(d):
        continue
    if turn.startswith(buyer):
        buyer_turns.append(turn[7:])
    else:
        seller_turns.append(turn[8:])

seller_turns
```

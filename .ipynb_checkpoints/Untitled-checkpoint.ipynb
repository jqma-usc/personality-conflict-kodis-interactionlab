```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming the original DataFrame 'df' is already loaded

# Filtering the DataFrame for a specific 'kodis-id'
specific_kodis_id = 2523  # Replace with the desired 'kodis-id'
filtered_df = df[df['kodis-id'] == specific_kodis_id]

# Further filtering for buyers and sellers
buyer_df = filtered_df[filtered_df['buyer or seller'] == 'buyer']
# seller_df = filtered_df[filtered_df['buyer or seller'] == 'seller']

# Define the columns for the plot
x_col = ''
y_col = 'Clout'

# Create the scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(buyer_df[x_col], buyer_df[y_col], label='Buyer', color='blue', alpha=0.7)
# plt.scatter(seller_df[x_col], seller_df[y_col], label='Seller', color='red', alpha=0.7)

plt.xlabel(x_col)
plt.ylabel(y_col)
plt.title(f'Scatter Plot for kodis-id {specific_kodis_id} vs {y_col}')
plt.legend()

# Adjust the y-axis limits according to the data range with a margin
if not filtered_df.empty:
    y_min = buyer_df[y_col].min()
    y_max = buyer_df[y_col].max()
    print(y_max)
    margin = (y_max - y_min) * 0.1  # 10% margin on top and bottom
    plt.ylim(y_min - margin, y_max + margin)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Filter the DataFrame for a specific 'kodis-id'
specific_kodis_id = 2523
filtered_df = df[df['kodis-id'] == specific_kodis_id]

# 2. Separate buyer and seller rows
buyer_df = filtered_df[filtered_df['buyer or seller'] == 'buyer']
seller_df = filtered_df[filtered_df['buyer or seller'] == 'seller']

# 3. Decide which columns are your x and y
x_col = 'kodis-id'  # e.g. 0, 1, 2, 3, ...
y_col = 'Clout'                 # or 'word_count', whichever metric you're plotting

plt.figure(figsize=(8, 6))

# 4. Plot buyer data (red line)
plt.plot(
    buyer_df[x_col],
    buyer_df[y_col],
    color='red',
    marker='o',
    label='buyer'
)

# 5. Plot seller data (blue line)
plt.plot(
    seller_df[x_col],
    seller_df[y_col],
    color='blue',
    marker='o',
    label='seller'
)

# 6. Labeling
plt.title(f"Word count over time for KODIS conversation #{specific_kodis_id}")
plt.xlabel("Sliding window index")  # Adjust label text if needed
plt.ylabel("Word count")            # or "Clout", depending on your column meaning
plt.legend(loc='lower right')

# 7. (Optional) Adjust the y-axis so points aren’t squashed at the top/bottom
if not filtered_df.empty:
    y_min = filtered_df[y_col].min()
    y_max = filtered_df[y_col].max()
    margin = (y_max - y_min) * 0.1
    plt.ylim(y_min - margin, y_max + margin)

plt.tight_layout()
plt.show()
```
